# Simple Example

```python
from landfall import plot_points
```

```python
# define points data lists
latitudes = [27.8569685, 27.9113421, 27.9436236, 27.9533251]
longitudes = [-82.4927991, -82.4966344, -82.4789019, -82.4552228]
# plot points in all blue
plot_points(latitudes, longitudes)
```

# Or pass (Latitude, Longitude) tuples

```python
# define points data tuples
points = [
    (29.425788, -98.486167),
    (29.425383, -98.491538),
    (29.419795, -98.483968),
    (29.435552, -98.487176),
]
# alternate input format
plot_points(points)
```

# Flip coordinates (in case yours are in (longitude, latitude) order)

```python
# define flipped points tuples (longitude, latitude)
lon_lats = [
    (-98.486167, 29.425788),
    (-98.491538, 29.425383),
    (-98.483968, 29.419795),
    (-98.487176, 29.435552),
]
# quick fix if your coordinates are in the wrong order
plot_points(lon_lats, flip_coords=True)
```

# Pass color codes to colors argument

```python
# define color codes list
colors = ["#b33951", "#800080", "#f58a07", "#00FFFF"]
# specify the color of each point
plot_points(latitudes, longitudes, colors=colors)
```

# Pass color code to color argument

```python
# specify the single color of all points using color code string
plot_points(latitudes, longitudes, color="#00FFFF")
```

# Pass RGB tuples to colors argument

```python
# define RGB color tuples
colors = [(255, 0, 0), (0, 255, 0), (0, 0, 255), (255, 255, 0)]
# specify the color of each point using RGB tuples
plot_points(latitudes, longitudes, colors=colors)
```

# Pass RGB tuple to color argument

```python
# specify the single color of all points using RGB tuple
plot_points(latitudes, longitudes, color=(255, 255, 0))
```

# Pass staticmaps colors to colors argument

```python
import staticmaps

# define staticmaps colors list
colors = [staticmaps.BLUE, staticmaps.GREEN, staticmaps.RED, staticmaps.YELLOW]
# specify the color of each point using staticmaps colors
plot_points(latitudes, longitudes, colors=colors)
```

# Pass staticmaps color to color argument

```python
# specify the single color of all points using a staticmaps color
plot_points(latitudes, longitudes, color=staticmaps.BROWN)
```

# Pass 'wheel', 'random', or 'distinct' to colors argument

```python
# 'wheel' picks opposite colors on the color wheel for each point
plot_points(latitudes, longitudes, colors="wheel")
```

```python
# 'random' picks random colors for each point
plot_points(latitudes, longitudes, colors="random")
```

```python
# 'distinct' picks visually distinct colors for each point
plot_points(latitudes, longitudes, colors="distinct")
```

# Change point icon size (default is 10)

```python
plot_points(latitudes, longitudes, point_size=15)
```

# Change window size (default is (500, 400))

```python
plot_points(latitudes, longitudes, window_size=(300, 500))
```

# Zoom in or out (default is 0)

```python
# negative integers for zooming out
plot_points(latitudes, longitudes, zoom=-1)
```

# Set zoom (override automatic zoom level)

```python
plot_points(latitudes, longitudes, set_zoom=5)
```

# Pass unique ids for points and give them unique colors

```python
ids = [1, 1, 2, 2]
# pass 'wheel', 'random', or 'distinct' to id_colors
plot_points(latitudes, longitudes, ids=ids, id_colors="wheel")
```

```python
# or pass dict of specific id colors
id_colors = {1: staticmaps.RED, 2: staticmaps.GREEN}
plot_points(latitudes, longitudes, ids=ids, id_colors=id_colors)
```

# Provide your own tile provider specs

```python
# Choose a supported py-staticmaps tile provider.
tp = staticmaps.tile_provider_OSM

plot_points(latitudes, longitudes, tile_provider=tp)
```

# Provide your own staticmaps context

```python
from staticmaps import Context, Circle, create_latlng, Color

context = Context()
tblue = Color(0, 0, 255, 75)
for point in points:
    ll = create_latlng(*point)
    circle = Circle(ll, 1, fill_color=tblue)
    context.add_object(circle)

plot_points(
    points,
    context=context,
)
```
